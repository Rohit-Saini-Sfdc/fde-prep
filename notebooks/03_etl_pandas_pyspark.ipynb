```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, avg, count, window

# Build local PySpark session in Colab
spark = SparkSession.builder \
    .appName("FDE-ETL-Pipeline") \
    .master("local[*]") \
    .getOrCreate()

print("Spark Version:", spark.version)

# Create sample dataframe
data = [
    ("Engineering", "Alice", 120000),
    ("Engineering", "Bob", 110000),
    ("Sales", "Charlie", 95000),
    ("Sales", "David", 105000),
]
df = spark.createDataFrame(data, ["department", "name", "salary"])
df.show()

# Aggregation query
summary = df.groupBy("department").agg(avg("salary").alias("avg_salary"), count("name").alias("employee_count"))
summary.show()
```
