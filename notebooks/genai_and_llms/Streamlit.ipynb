What it does:

! tells Colab to run the line as a Linux bash command instead of Python.

wget downloads data from a web URL.

-q (quiet mode) silences the download progress output.

-O - (uppercase O followed by a dash) streams the downloaded output directly to your screen rather than saving it to a file.

ipv4.icanhazip.com returns the public IPv4 address of the Google Colab virtual machine.

Why it's needed: localtunnel protects its free tunnels with a security landing page. It requires the host machine's public IP as an access password before it lets you view the web app.

What it does:

streamlit run app.py: Starts the Streamlit local web server on its default port (8501).

& (background operator): Runs the Streamlit process in the background so the cell does not get blocked and can immediately execute the next command in the same line.

npx: Node.js package runner that downloads and executes an npm package without needing to permanently install it first.

localtunnel --port 8501: Creates a secure public tunnel from the internet to port 8501 inside Google's private virtual machine, printing a public URL (e.g., [https://xxxx.loca.lt](https://xxxx.loca.lt)).

Why it's needed: Google Colab runs in a private cloud container without exposed public ports. Standard localhost:8501 is trapped inside Google's server; localtunnel exposes it so your local browser can connect.

###Displaying Text & Formatting
st.header("Section Title") / st.subheader("Sub-section") — Renders hierarchical headings.

st.caption("Small grey text") — Displays small, muted text for footnotes or secondary notes.

st.code("x = [1, 2, 3]", language="python") — Renders a syntax-highlighted code block with a one-click copy button.

st.latex(r''' e^{i\pi} + 1 = 0 ''') — Renders mathematical formulas using LaTeX syntax.

st.divider() — Inserts a clean horizontal rule.

###User Inputs & Controls
st.text_input("Name", value="Default") / st.text_area("Bio") — Captures single-line or multi-line text.

st.number_input("Age", min_value=0, max_value=120, step=1) — Numeric field with increment/decrement steppers.

st.checkbox("I agree") — Returns a Boolean (True/False).

st.radio("Pick one", ["Option A", "Option B", "Option C"]) — Radio button list for mutually exclusive choices.

st.multiselect("Pick tags", ["Python", "SQL", "Docker"]) — Dropdown allowing multiple items to be selected as a list.

st.file_uploader("Upload CSV", type=["csv", "xlsx"]) — File upload widget that returns a buffer readable by pandas.

st.color_picker("Pick a color", "#00f900") — Interactive color wheel returning a hex string.

###Layouts & Containers
st.columns([2, 1]) — Splits the screen into side-by-side columns with custom proportional widths:

Python
col1, col2 = st.columns([2, 1])
with col1:
    st.write("Main wide area")
with col2:
    st.write("Sidebar-like narrow column")
st.tabs(["Tab 1", "Tab 2"]) — Separates content into clickable tab views:

Python
tab1, tab2 = st.tabs(["Overview", "Raw Data"])
with tab1:
    st.write("Summary charts")
st.expander("Click to view details") — A collapsible container that saves vertical space.

st.container(height=300) — A scrollable or fixed-height bounding box for dynamic content.

###Status, Alerts & Feedback
st.success("Done!") / st.error("Failed!") / st.warning("Warning!") / st.info("Note") — Styled alert boxes.

st.toast("Saved successfully!", icon="✅") — Displays a temporary popup notification in the bottom-right corner.

st.spinner("Crunching numbers...") — Displays a temporary loading spinner while a code block executes:

Python
with st.spinner("Processing..."):
    time.sleep(2)
st.progress(75) — Renders a progress bar (accepts values from 0 to 100 or 0.0 to 1.0).

st.balloons() / st.snow() — Full-screen celebration animations.

###Interactive Chat & AI UI
st.chat_message("user") / st.chat_message("assistant") — Pre-styled speech bubbles for conversational interfaces:

Python
with st.chat_message("assistant"):
    st.write("How can I help you today?")
st.chat_input("Ask a question...") — Pins a sticky chat prompt bar at the bottom of the viewport.

###Media & Charting
st.audio("audio.mp3") / st.video("video.mp4") — Native media players.

st.bar_chart(df) / st.area_chart(df) / st.scatter_chart(df) — Built-in interactive charts powered by Altair.

st.pyplot(fig) / st.plotly_chart(fig) — Directly renders charts created in Matplotlib, Seaborn, or Plotly.

```python
%%writefile app.py
import streamlit as st

st.set_page_config(page_title="Colab Streamlit", layout="wide")
st.title("Streamlit on Google Colab 🚀")
st.write("If you can see this, the connection works!")

val = st.slider("Select a value", 0, 100, 25)
st.write(f"Current Value: {val}")
```

```python
import subprocess
import time

# 1. Kill any existing instances
subprocess.run(["pkill", "-9", "streamlit"], stderr=subprocess.DEVNULL)

# 2. Spawn Streamlit in an independent, fully detached background process
cmd = [
    "streamlit", "run", "app.py",
    "--server.port", "8501",
    "--server.headless", "true",
    "--server.enableCORS", "false",
    "--server.enableXsrfProtection", "false",
    "--browser.gatherUsageStats", "false"
]
subprocess.Popen(cmd, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)

# 3. Wait briefly for the server to bind to port 8501
time.sleep(3)
print("Streamlit is running in the background!")
```
